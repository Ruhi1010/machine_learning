```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
x = np.linspace(0, 5, 11)
print(x)
y = x ** 2
print(y)
```

Plots in Matplotlib

```python
plt.plot(x, y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
plt.legend(['y = x^2'])
plt.show()
```

Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x**3, y**2, 'r')
plt.subplot(2,2,2)
plt.plot(x ** 2, x ** 2, 'b')
plt.subplot(2,2,3)
plt.plot(y, x, 'g')
plt.subplot(2,2,4)
plt.plot(x, y, 'r')
```

Object Oriented Way

```python
fig = plt.figure(figsize=(6, 8), dpi=100)
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x, y, 'r')
axis2 = fig.add_axes([0.3, 0.8, 0.3, 0.3])
axis2.plot(y, x, 'b')
```

Types of Plot

Saving the plot

```python
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.plot(x, y)
plt.savefig('test.png', dpi=200)
```
