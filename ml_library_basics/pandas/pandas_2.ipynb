````python
# 1. Calculate overall mean and standard deviation
mean_salary = df["salary"].mean()
std_salary = df["salary"].std()

# 2. Define upper and lower wave boundaries
lower_bound = mean_salary - std_salary
upper_bound = mean_salary + std_salary

# 3. Create the plot
plt.figure(figsize=(12, 6))

# Plot the individual salary wave (assuming DataFrame is sorted or sequential)
plt.plot(
    df.index, df["salary"], color="navy", alpha=0.6, label="Employee Salary"
)

# Plot the center static mean line
plt.axhline(mean_salary, color="blue", linestyle="--", label="Mean Salary")

# Shade the ±1 Standard Deviation "wave channel"
plt.fill_between(
    df.index,
    lower_bound,
    upper_bound,
    color="blue",
    alpha=0.15,
    label="±1 Std Dev Range",
)

# 4. Labels and formatting
plt.xlabel("Index / Employee sequence")
plt.ylabel("Salary")
plt.title("Salary Wave Plot with Standard Deviation Band")
plt.legend(loc="upper left")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python

# 1. Calculate the mean and standard deviation
mean_val = df["salary"].median()
std_val = df["salary"].std()

# 2. Generate the continuous distribution curve (KDE)
plt.figure(figsize=(10, 6))
ax = df["salary"].plot(kind="density", color="blue", linewidth=2, label="Salary Distribution")

# 3. Extract the calculated curve coordinates to shade it precisely
line = ax.get_lines()[0]
x_data, y_data = line.get_xdata(), line.get_ydata()

# 4. Create masks for the standard deviation regions
one_std_mask = (x_data >= mean_val - std_val) & (x_data <= mean_val + std_val)

# 5. Fill the standard deviation "wave" regions under the curve
plt.fill_between(
    x_data,
    y_data,
    where=one_std_mask,
    color="blue",
    alpha=0.2,
    label="±1 Std Dev Range",
)

# 6. Add reference lines for context
plt.axvline(mean_val, color="blue", linestyle="dashed", linewidth=1.5, label=f"Mean: {mean_val:,.0f}")
plt.axvline(mean_val - std_val, color="red", linestyle="dotted", linewidth=1.5, label="-1 SD")
plt.axvline(mean_val + std_val, color="red", linestyle="dotted", linewidth=1.5, label="+1 SD")

# 7. Labels and formatting
plt.title("Salary Distribution Wave with Standard Deviation Band", fontsize=14)
plt.xlabel("Salary")
plt.ylabel("Density")
plt.xlim(df["salary"].min() - std_val, df["salary"].max() + std_val)
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()
````

#### There have a case about this null value fill up that is if my salary column is store `string` type data but there have some `null` values in the column, so there we use to covert the `string` data to `numeric` data first after the conversion we will fill the data by `mean` or `median` 

#### Here we need to conversion for when we use `mean` `median` to fill the `null` space or rows there must need to have `numeric` data to fill because these functions are use for `numeric` data, so this reason we need to convert the data into `numeric` if the datatype is not in `numeric`

#### as example:
```python
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['salary'] = df['salary'].fillna(df['salary'].median())
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0 or 1 or 2])
```

````python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'A': [21, np.nan, 22, 4, np.nan],
    'B': [31, 32, 33, 34, 35],
    'C': [np.nan, np.nan, np.nan, np.nan, np.nan],
    'D': [46, 47, np.nan, 48, 49]
})

print("Original DataFrame:")
print(df)

# Count NaN values
print("\nNaN values in each column:")
print(df.isna().sum())

print("\nTotal NaN values:")
print(df.isna().sum().sum())

# Forward Fill
print("\nForward Fill:")
print(df.ffill())

# Backward Fill
print("\nBackward Fill:")
print(df.bfill())

# Fill with 0
print("\nFill NaN with 0:")
print(df.fillna(0))
````
