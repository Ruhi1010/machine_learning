```python
import plotly.express as px

tip_by_day = tps.groupby('day', as_index=False)['tip'].mean()

fig = px.bar(
    tip_by_day,
    x='day',
    y='tip'
)

fig.show()
```

```python
import plotly.express as px

fig = px.scatter(
    tps,
    x='total_bill',
    y='tip'
)

fig.show()
```

```python
import plotly.express as px

fig = px.box(
    tps,
    x='total_bill',
    y='day'
)

fig.show()
```
