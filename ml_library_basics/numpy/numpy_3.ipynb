# Advance operations with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restaurant_id, 2022, 2023, 2024, 2025]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 180000], # Beijing Bites
    [3, 200000, 210000, 230000, 260000], # Pizza Hut
    [4, 170000, 190000, 210000, 240000], # Burger King
    [5, 130000, 150000, 170000, 200000]  # Cha and Chill
])
```

```python
print("==== food panda sales analysis ====")
```

```python
print("Sales data shape: \n", sales_data.shape)
```

```python
print("Sample data for first three restaurants: \n", sales_data[:,1:])
```

```python
# total sales per year
print("Total sales per year: \n", np.sum(sales_data[:,1:], axis=0))
```

```python
# Minimum sales per restaurant
min_sales_restaurant = np.min(sales_data[:,1:], axis=1)
print("Minimum sales per restaurant: \n", min_sales_restaurant)
```

```python
# Maximum sales per year
max_sales_year = np.max(sales_data[:,1:], axis=0)
print("Maximum sales per year: \n", max_sales_year)
```

```python
# Average sales per restaurant
avg_sales_restaurant = np.mean(sales_data[:,1:], axis=1)
print("Average sales per restaurant: \n", avg_sales_restaurant)
```

```python
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print("Cumulative sales per restaurant: \n", cumsum)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Sales accross all Restaurants")
plt.xlabel("Years (2022-2025)")
plt.ylabel("Average Cumulative Sales")
plt.grid(True)
plt.show()
```
