```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
```

```python
df['size'].unique()
```

distplot/histplot

```python
sns.displot(df['total_bill'], bins = 20, kde = True)
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], bins = 20, kde = True)
plt.subplot(1, 2, 2)
sns.histplot(df['tip'], bins = 20, kde = True)
```

jointplot

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'hex')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'kde')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'scatter')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'reg')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'resid')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind = 'hist')
```

pairplot

```python
sns.pairplot(df, hue='size', palette='rainbow')
```

rugplot

```python
sns.rugplot(df['total_bill'])
```

countplot

```python
sns.countplot(x = df['sex'], hue=df['smoker'])
```

```python
sns.barplot(x = df['sex'], y = df['total_bill'], hue=df['smoker'])
```

```python
import numpy as np
sns.barplot(x = df['sex'], y = df['tip'], estimator=np.sum)
```

boxplot

```python
sns.boxplot(data = df,y = 'tip', x = 'day', palette='rainbow')
```

violinplot

```python
sns.violinplot(data = df,y = 'tip', x = 'day', palette='rainbow')
```

stripplot

```python
sns.stripplot(data = df, x = 'day', y = 'total_bill')
```

swarmplot

```python
sns.swarmplot(data = df, x = 'day', y = 'total_bill')
```

```python
sns.violinplot(data = df, x = 'tip', y = 'day')
sns.swarmplot(data = df, x = 'tip', y = 'day')
```

Heat map

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tipscorr = tips[['total_bill', 'tip', 'size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(), annot=True)
```

Cluster map

```python
sns.clustermap(tipscorr.corr(), annot=True)
```

pivot table heat map

```python
pivot_flights = flights.pivot_table(index='month', columns='year', values='passengers')
```

```python
pivot_flights
```

```python
sns.heatmap(pivot_flights)
```

lmplot() -> regression plot

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=tips)
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=tips, hue='sex')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=tips, hue='sex', palette='coolwarm')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data=tips, hue='sex', palette='coolwarm', markers=['o', 'v'], scatter_kws={'s': 100})
```
